```python
# Imports
from langgraph.graph import StateGraph, START, END
from langchain_google_genai import ChatGoogleGenerativeAI
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash"
)
```

```python
class LLMBlogState(TypedDict):
    title: str
    outline: str
    blog: str
```

```python
def get_blog(state: LLMBlogState) -> LLMBlogState:
    title = state["title"]
    outline = state["outline"]

    prompt = f"Write a blog about {title} with these outlines: {outline}"
    
    blog = llm.invoke(prompt).content
    state["blog"] = blog

    return state
```

```python
def get_outline(state: LLMBlogState) -> LLMBlogState:
    title = state["title"]
    prompt = f"Write few outline about {title}."

    outline = llm.invoke(prompt).content
    state["outline"] = outline

    return state
```

```python
graph = StateGraph(LLMBlogState)

graph.add_node("get_outline",get_outline)
graph.add_node("get_blog",get_blog)

graph.add_edge(START,"get_outline")
graph.add_edge("get_outline","get_blog")
graph.add_edge("get_blog",END)

workflow = graph.compile()

initial_state = {"title":"IPL in todays era"}
final_state = workflow.invoke(initial_state)
print(final_state)
```
